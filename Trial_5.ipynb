```python
df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")

print(df.columns)
```

```python
#renaming columns names
df = df.rename(columns={"gross_margin_percentage": "gross_margin",
                       "unit_price_(ghs)": "unit_price",
                       "tax_5%": "tax"})
print(df.columns)
```

```python
#before
print("Type before", df['unit_price'].dtype)
print(df['unit_price'].head(3))
```

```python
#remove ghs from it
df['unit_price'] = df['unit_price'].str.replace("GHS", "").str.strip()

print(df['unit_price'])
```

```python
#convert to a proper numbrer
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")

print("Type after", df['unit_price'].dtype)
print(df['unit_price'].head())



```

```python
df['date'].head()
```

```python
df['date'] = pd.to_datetime(df['date'], format="mixed")

print("after", df['date'].dtype)
print(df['date'].head(6))
```

```python
#extract months and days from the date
df["month"]=df['date'].dt.month
df["day"]=df['date'].dt.day
df["year"]=df["date"].dt.year
df[["date", "month", "year", "day"]].head(5)
```

```python
df.dtypes
```

```python
print(df.isnull().sum())
```

```python
missing_pct = (df.isnull().sum()/len(df)*100).round(1)

print(missing_pct)
```

```python
# filling rating to a complete number
df['rating'] = df['rating'].fillna(df['rating'].mean())

df['total'] = df['total'].fillna(df['total'].mean())

print("Missing in rating", df["rating"].isnull().sum())
print("Missing in total", df["total"].isnull().sum())
```

```python
# filling text
df['branch']= df['branch'].fillna("unknown")
df['customer_type']= df['customer_type'].fillna("unknown")
df['gender']= df['gender'].fillna("unknown")

print("Missing in branch", df["branch"].isnull().sum())
print("Missing in customer_type", df["customer_type"].isnull().sum())
print("Missing in gender", df["gender"].isnull().sum())
```

Groupby and aggregation


```python
# total revenue per branch
df.groupby("branch")["total"].sum().round(2)
```

```python
# average rating per branch
df.groupby("branch")["rating"].mean().round(2)
```

```python
# number of transaction per product line
df.groupby("product_line")["invoice_id"].count()
```

```python
# count, total and average per product_line
df.groupby("product_line")["total"].agg(["count", "sum", "mean"]).round(2)
```

```python
df.head(5)
```

```python
# group by two columns
df.groupby(["branch", "customer_type"])["total"].sum().round(2)
```

```python
df["customer_type"] = df["customer_type"].str.capitalize()
df["customer_type"].unique()
```

Ranking

```python
# rank product line by total revenue
product_revenue=(df.groupby("product_line")["total"].sum().round().reset_index().rename(columns={"total": "total_revenue"}))

product_revenue.sort_values("total_revenue", ascending=False)
print(product_revenue)
```

```python
product_revenue = (df.groupby("product_line")["total"].sum().round(2))

product_rank = product_revenue.rank(ascending=False)

print(product_rank)
```

Which product has the highet revenue

```python
highest_revenue_product = (
    df.groupby("product_line")["total"].sum()
    .round(2)
    .reset_index()
    .rename(columns={"total": "total_revenue"})
)

highest_revenue_product = highest_revenue_product.sort_values(
    by="total_revenue", ascending=False
)

print(highest_revenue_product)
```

Average revenue on which customer_type

```python
# total + mean + sort
customer_average_revenue = (df.groupby("customer_type")["total"].mean().round(2)
                            .reset_index()
                            .rename(columns={"total": "average_revenue"}))

customer_average_revenue = customer_average_revenue.sort_values(by="average_revenue", ascending=False)

print(customer_average_revenue)
```
